AI CHATBOT WITHOUT MEMORY

```python
from groq import Groq # - importing lib
#Creating the client
client = Groq(api_key="YOUR_API_KEY")
#Infinite loop
while True:
  user_input = input("You:")
  #calling the ai model
  response = client.chat.completions.create(
      model = "openai/gpt-oss-20b", # selecting the model
      messages=[ #Sending the message
          {"role":"user","content":user_input}
      ]
  )
  # we have to get the reply
  reply = response.choices[0].message.content
  # printing the reply
  print("AI:",reply)
```

Chatbot with a memory

```python
from groq import Groq
client = Groq(api_key="YOUR_API_KEY")
messages = []

while True:
  user_input = input("You:")
  messages.append({"role":"user","content":user_input})
  response = client.chat.completions.create(
      model = "openai/gpt-oss-20b",
      messages=messages
  )
  reply = response.choices[0].message.content
  print("AI:",reply)
```
