Linguistic Analyzer - Understanding Sentence Structure

we will Examine:



1.   Word formation through simple morphology rules
2.  Sentence structure through basic word classif
3. Word meanings through a small dict
4. Contextual interpretations through eg;





```python
import re
```

2)taking sentence i/p

```python
sentence = input("Enter a sentence: ")

print("Original Sentence:", sentence)
```

3)Basic word tokenization

```python
words = re.findall(r'\b\w+\b', sentence.lower())

print("Tokens:", words)
```

4)basic morphology

```python
suffixes = ["ing", "ed", "ness", "ly", "s"]

for word in words:

    root = word

    for suffix in suffixes:

        if word.endswith(suffix) and len(word) > len(suffix) + 2:
            root = word[:-len(suffix)]
            break

    print(word, "->", root)
```

5)Basic Syntax analysis

```python
for word in words:

    if word in ["the", "a", "an"]:
        print(word, "-> Determiner")

    elif word in ["is", "are", "was", "were", "am"]:
        print(word, "-> Verb")

    elif word.endswith("ing") or word.endswith("ed"):
        print(word, "-> Possible Verb")

    else:
        print(word, "-> Other")
```

6. Basic semantic analysis

```python
word_meanings = {
    "bank": ["Financial institution", "River side"],
    "apple": ["Fruit", "Technology company"],
    "python": ["Programming language", "Snake"]
}

for word in words:

    if word in word_meanings:
        print(word, "can mean:", word_meanings[word])
```

7. Basic Pragmatic analysis

```python
pragmatic_examples = {
    "it's hot in here": "Possible request to switch on a fan",
    "can you open the door": "Usually an indirect request",
    "wow you're early": "Could be sarcasm depending on context"
}

normalized = sentence.lower().strip()

if normalized in pragmatic_examples:
    print("Possible intended meaning:",
          pragmatic_examples[normalized])
else:
    print("Context is required for interpretation.")
```

8. Final Linguistic report

```python
print("\n========== LINGUISTIC ANALYZER ==========")

print("Original Sentence:", sentence)

print("\nMorphological Tokens:")
print(words)

print("\nNumber of Words:", len(words))

print("\nPossible Linguistic Levels:")
print("1. Morphology - Word formation")
print("2. Syntax - Sentence structure")
print("3. Semantics - Word and sentence meaning")
print("4. Pragmatics - Contextual interpretation")

print("=========================================")
```
