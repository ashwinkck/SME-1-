1. Importing

```python
import re #re - Regular expressions
from collections import Counter # Counter - Counts how frequently individual elements occur
```

# 2. Input text

```python
text = """Natural Language processing is a branch of Aritificial Intelligence.
NLP helps computers process human language.
I love learning NLP because NLP is interesting!"""
```

3. Converting text to lowercase

```python
clean_text = text.lower()
print(clean_text)
```

4. Remove Punctuation

```python
clean_text = re.sub(r'[^a-zA-Z\s]','',clean_text)
print(clean_text)
```

5. Tokenization

```python
words = clean_text.split()
print(words)
```

6. Count total words

```python
total_words = len(words)
print("Total words:",total_words)
```

7. Calculate word Frequency

```python
word_frequency = Counter(words)
print(word_frequency)
```

8. Displaying the top five words

```python
top_words = word_frequency.most_common(5)
print("Top 5 frequent words:")
for word, count in top_words:
  print(word,":",count)
```

9. counting sentences

```python
sentences = re.split(r'[.!?]+', text)
sentences = [s.strip() for s in sentences if s.strip()]
print("Total sentences:", len(sentences))
```

10. final NLP report

```python
print("============== NLP TEXT ANALYZER===========")
print("Total Words:",total_words)
print("Total Sentences:", len(sentences))

print("\nMost Frequent Words:")

for word, count in top_words:
  print(word,":",count)

print("=================================================================")
```
