2. Importing modules

```python
import nltk

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

3. Downloading the required language resources

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
```

4.Initializing the models

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

5.Performing Stemming

```python
words = [
    "playing",
    "studies",
    "connected",
    "happiness",
    "running",
    "cars"
]
print("STEMMING RESULTS")

for word in words:
  stem = stemmer.stem(word)
  print(word,"->",stem)
```

6. Performing Lemmatization

```python
words = [
    "playing",
    "studies",
    "cars",
    "mice",
    "running"
]
print("LEMMATIZATION RESULTS")

for word in words:
  lemma = lemmatizer.lemmatize(word)
  print(word,"->",lemma)
```

7.Lemmatization with POS info (part of speech)

```python
print("LEMMATIZATION WITH POS")
print("playing ->",lemmatizer.lemmatize("playing",pos="v"))
print("running ->",lemmatizer.lemmatize("running",pos="v"))
print("better - >",lemmatizer.lemmatize("better",pos="a"))
print("mice - >",lemmatizer.lemmatize("mice",pos="n"))
```

8. Build a simple morphological Analyzer

```python
def morphological_analyzer(word):
  print("\nWord:",word)

  #Stemming
  stem = stemmer.stem(word)

  #Lemmatization
  lemma = lemmatizer.lemmatize(word,pos="v")
  print("Stem:",stem)
  print("Lemma:",lemma)

morphological_analyzer("playing")
morphological_analyzer("running")
morphological_analyzer("studies")
morphological_analyzer("connected")
```

9. Basic prefix and suffix identification

```python
prefixes = ["un","re","dis","im","pre"]

suffixes = ["ing","ed","ness","ful","less","ly"]

def identify_affixes(word):
  found_prefix = "None"
  found_suffix = "None"

  for prefix in prefixes:
    if word.startswith(prefix):
      found_prefix = prefix
      break
  for suffix in suffixes:
    if word.endswith(suffix):
      found_suffix = suffix
      break
  print("Word:",word)
  print("Prefix:",found_prefix)
  print("Suffix:",found_suffix)

identify_affixes("unhappiness")
identify_affixes("replaying")
identify_affixes("carefully")
```
