2 - Loading the NLP Model

```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("NLP model loaded successfully")
```

3 - Preparinmg the sample test

```python
text = """
Satya Nadella is the CEO of Microsoft.
He was born in Hyderabad,India.
Microsoft was founded in 1975.
The company develops software and AI techs
"""
doc = nlp(text)
print(doc)
```

4 - POS Taggging

```python
print("POS TAGGING")
print("-"*40)

for token in doc:
  print(
      token.text,
      "->",
      token.pos_,
      "|",
      token.tag_
  )
```

5 - Chunking

```python
print("NOUN PHRASES")
print("-"*40)

for chunk in doc.noun_chunks:
  print(chunk.text)
```

6 - NER -Named entity Recognition

```python
print("NER")
print("-"*40)

for ent in doc.ents:
  print(
      ent.text,
      "->",
      ent.label_
  )
```

7 - Converting entitites into Structured info

```python
entities = {
    "PERSON":[],
    "ORG":[],
    "GPE":[],
    "DATE":[],
    "MONEY":[],
    "LOC":[]
}

for ent in doc.ents:
  if ent.label_ in entities:
    entities[ent.label_].append(ent.text)

for category, values in entities.items():
  print(f"\n{category}:")
  for value in values:
    print("-",value)
```

Complete Code

```python

# ==========================================
# DAY 05: INTELLIGENT TEXT INFORMATION EXTRACTOR
# ==========================================

import spacy

# Load pretrained NLP model
nlp = spacy.load("en_core_web_sm")


def extract_information(text):

    # Process input text
    doc = nlp(text)

    print("\n========== INPUT TEXT ==========")
    print(text)

    # 1. POS TAGGING
    print("\n========== POS TAGGING ==========")

    for token in doc:
        if not token.is_space and not token.is_punct:
            print(f"{token.text:20} {token.pos_}")

    # 2. NOUN PHRASES
    print("\n========== NOUN PHRASES ==========")

    for chunk in doc.noun_chunks:
        print("-", chunk.text)

    # 3. NAMED ENTITY RECOGNITION
    print("\n========== NAMED ENTITIES ==========")

    entities = {}

    for ent in doc.ents:

        label = ent.label_

        if label not in entities:
            entities[label] = []

        entities[label].append(ent.text)

        print(f"{ent.text:25} {label}")

    # 4. STRUCTURED OUTPUT
    print("\n========== STRUCTURED INFORMATION ==========")

    for label, values in entities.items():
        print(f"{label}: {values}")

    return entities


# Sample input
sample_text = """
Sundar Pichai is the CEO of Google.
He works in California, United States.
Google introduced several AI technologies in 2024.
"""

# Run application
result = extract_information(sample_text)
```
